###Configuring the Azure Data Lake paths

```python
source_path = "abfss://source@stecomdev0011.dfs.core.windows.net/customers/"
schema_path = "abfss://bronze@stecomdev0011.dfs.core.windows.net/_schemas/customers/"
checkpoint_path = "abfss://bronze@stecomdev0011.dfs.core.windows.net/_checkpoints/customers/"
bronze_output_path = "abfss://bronze@stecomdev0011.dfs.core.windows.net/customers/"
```

## Reading data into Dataframe

```python
df_customers = (
    spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "parquet")
        .option("cloudFiles.schemaLocation", schema_path)
        .load(source_path)
)
```

## Writing data into Silver Layer.

```python
(
    df_customers.writeStream
        .format("parquet")
        .option("checkpointLocation", checkpoint_path)
        .outputMode("append")
        .trigger(availableNow=True)
        .start(bronze_output_path)
)
```
