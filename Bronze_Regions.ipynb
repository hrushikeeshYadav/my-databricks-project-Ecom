```python
# Ingest region data into the Bronze Delta table
source_path = "abfss://source@stecomdev0011.dfs.core.windows.net/regions/"
schema_path = "abfss://bronze@stecomdev0011.dfs.core.windows.net/_schemas/regions/"
checkpoint_path = "abfss://bronze@stecomdev0011.dfs.core.windows.net/_checkpoints/regions/"

df_regions = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "parquet")
    .option("cloudFiles.schemaLocation", schema_path)
    .load(source_path)
)

query_regions = (
    df_regions.writeStream
    .format("delta")
    .option("checkpointLocation", checkpoint_path)
    .trigger(availableNow=True)
    .toTable("ecom_dev.bronze.regions")
)

query_regions.awaitTermination()
```
