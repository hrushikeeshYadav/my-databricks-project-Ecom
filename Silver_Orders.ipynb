```python
# Transform and write order data to Silver Delta table
bronze_df = spark.table("ecom_dev.bronze.orders")

silver_df = (
    bronze_df
    .dropDuplicates(["order_id"])
    .where("customer_id IS NOT NULL AND product_id IS NOT NULL")
)

(
    silver_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("ecom_dev.silver.orders")
)
```
