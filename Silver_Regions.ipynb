```python
%python
# Transform and write region data to Silver Delta table
bronze_df = spark.table("ecom_dev.bronze.regions")

silver_df = (
    bronze_df
    .dropDuplicates(["region_id"])
    .where("region IS NOT NULL")
)

(
    silver_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("ecom_dev.silver.regions")
)
```
