```python
# Transform and write customer data to Silver Delta table
bronze_df = spark.table("ecom_dev.bronze.customer")

silver_df = (
    bronze_df
    .dropDuplicates(["customer_id"])
    .where("email IS NOT NULL")
)

(
    silver_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("ecom_dev.silver.customers")
)
```

```python
# Read and display data from the Silver customers table
df_silver_customers = spark.table("ecom_dev.silver.customers")
display(df_silver_customers)
```
