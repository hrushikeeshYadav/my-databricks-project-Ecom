```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number

df_dim_customers = spark.table("ecom_dev.silver.customers")

w = Window.orderBy("customer_id")

gold_dim = (
    df_dim_customers
    .dropDuplicates(["customer_id"])
    .withColumn("customer_key", row_number().over(w))
)

# Reorder columns so that customer_key is first
cols = ["customer_key"] + [col for col in gold_dim.columns if col != "customer_key"]
gold_dim = gold_dim.select(cols)

(
    gold_dim.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("ecom_dev.gold.dimension")
)
```
