```python
# Transform and write product data to Silver Delta table
bronze_df = spark.table("ecom_dev.bronze.products")

silver_df = (
    bronze_df
    .dropDuplicates(["product_id"])
    .where("product_name IS NOT NULL")
)

(
    silver_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("ecom_dev.silver.products")
)
```
